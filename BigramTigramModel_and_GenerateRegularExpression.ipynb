```python
import nltk
from nltk import bigrams, trigrams
from nltk.tokenize import word_tokenize
import re
```

```python
text = "I love natural language processing and I love machine learning"
tokens = word_tokenize(text)
```

```python
bigram_list = list(bigrams(tokens))
trigram_list = list(trigrams(tokens))
```

```python
print("Bigrams:", bigram_list)
print("Trigrams:", trigram_list)
```

```python
# Simple bigram frequency model
from collections import defaultdict
```

```python
model = defaultdict(list)
for w1, w2 in bigram_list:
    model[w1].append(w2)
```

```python
print("\nBigram model (word -> possible next words):")
for w, v in model.items():
    print(w, "->", v)
```

```python
# Regular expression example: extract words that strat with a vowel
pattern = r'\b[aeiouAEIOU]\W*\b'
matches = re.findall(pattern, text)
print("\nWords strating with a vowel:", matches)
```
